```python
# --- Função para criar uma imagem PPM de teste ---
def criar_ppm_teste(nome="entrada.ppm"):
    conteudo = """P3
# Imagem teste gerada automaticamente
2 2
255
255 0 0   0 255 0
0 0 255   200 200 200
"""
    with open(nome, "w") as f:
        f.write(conteudo)
    print(f"✅ Arquivo {nome} criado com sucesso!")

# --- Função para ler PPM ---
def ler_ppm(arquivo):
    with open(arquivo, 'r') as f:
        tipo = f.readline().strip()
        if tipo != 'P3':
            raise ValueError("Apenas imagens PPM P3 suportadas.")
        f.readline()  
        largura, altura = map(int, f.readline().split())
        max_val = int(f.readline().strip())
        pixels = list(map(int, f.read().split()))
        
    imagem = []
    k = 0
    for i in range(altura):
        linha = []
        for j in range(largura):
            r, g, b = pixels[k:k+3]
            linha.append((r, g, b))
            k += 3
        imagem.append(linha)
    return imagem, largura, altura, max_val

# --- Função para salvar PPM ---
def salvar_ppm(nome, imagem, largura, altura):
    with open(nome, 'w') as f:
        f.write("P3\n")
        f.write(f"{largura} {altura}\n255\n")
        for linha in imagem:
            for r, g, b in linha:
                f.write(f"{r} {g} {b} ")
            f.write("\n")

# --- Funções de processamento ---
def rgb_para_cinza(r, g, b):
    return int(0.299*r + 0.587*g + 0.114*b)

def binarizar_pixel(c, limiar=128):
    return 255 if c > limiar else 0

# --- Função principal ---
def processar_imagem(entrada, saida_cinza, saida_bin):
    img, larg, alt, _ = ler_ppm(entrada)
    img_cinza = [[(c, c, c) for (r,g,b) in linha for c in [rgb_para_cinza(r,g,b)]] for linha in img]
    img_bin = [[(p, p, p) for (r,g,b) in linha for p in [binarizar_pixel(rgb_para_cinza(r,g,b))]] for linha in img]
    salvar_ppm(saida_cinza, img_cinza, larg, alt)
    salvar_ppm(saida_bin, img_bin, larg, alt)
    print("✅ Processamento concluído! Arquivos gerados:", saida_cinza, "e", saida_bin)

# --- Criar imagem de teste e processar ---
criar_ppm_teste()
processar_imagem("entrada.ppm", "saida_cinza.ppm", "saida_bin.ppm")
```
